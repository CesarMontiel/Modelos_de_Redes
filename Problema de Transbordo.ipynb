# Introducción

El Problema de Transbordo es una extensión del Problema de Transporte.

En este modelo la mercancía puede pasar por uno o varios nodos intermedios antes de llegar al cliente final.

Los nodos intermedios representan generalmente:

- Centros de distribución
- Puertos
- Aeropuertos
- Terminales ferroviarias
- Centros de consolidación

El objetivo consiste en minimizar el costo total de transporte respetando las capacidades de oferta y demanda.

```python
# Si PuLP no está instalado ejecutar:

# !pip install pulp
```

```python
from pulp import *
```

# Definición del problema

Una empresa cuenta con:

- Planta A (40 toneladas)
- Planta B (60 toneladas)

Toda la mercancía debe enviarse primero al Centro de Distribución (CD).

Posteriormente el CD abastece a:

- Cliente 1 (45 toneladas)
- Cliente 2 (55 toneladas)

Los costos son:

Planta A → CD = \$4

Planta B → CD = \$3

CD → Cliente 1 = \$5

CD → Cliente 2 = \$7

```python
oferta = {

    "PlantaA":40,

    "PlantaB":60

}

demanda = {

    "Cliente1":45,

    "Cliente2":55

}

costos = {

    ("PlantaA","CD"):4,

    ("PlantaB","CD"):3,

    ("CD","Cliente1"):5,

    ("CD","Cliente2"):7

}
```

# Creación del modelo

Se define un problema de minimización.

```python
modelo = LpProblem(
    "Problema_Transbordo",
    LpMinimize
)
```

# Variables de decisión

Las variables representan la cantidad enviada entre cada par de nodos.

Por ejemplo:

x_ACD

Cantidad enviada desde Planta A hacia el Centro de Distribución.

```python
x_ACD = LpVariable(
    "PlantaA_CD",
    lowBound=0
)

x_BCD = LpVariable(
    "PlantaB_CD",
    lowBound=0
)

x_CD1 = LpVariable(
    "CD_Cliente1",
    lowBound=0
)

x_CD2 = LpVariable(
    "CD_Cliente2",
    lowBound=0
)
```

# Función Objetivo

Se busca minimizar el costo total del transporte.

```python
modelo += (

    4*x_ACD +

    3*x_BCD +

    5*x_CD1 +

    7*x_CD2

)
```

# Restricciones de Oferta

Cada planta únicamente puede enviar la cantidad disponible.

```python
modelo += x_ACD == 40

modelo += x_BCD == 60
```

# Conservación del Flujo

Toda la mercancía que llega al Centro de Distribución debe salir hacia los clientes.

```python
modelo += (

    x_ACD +

    x_BCD

    ==

    x_CD1 +

    x_CD2

)
```

# Restricciones de Demanda

Cada cliente debe recibir exactamente la cantidad solicitada.

```python
modelo += x_CD1 == 45

modelo += x_CD2 == 55
```

# Resolver el modelo

```python
modelo.solve()
```

# Estado de la solución

```python
print(LpStatus[modelo.status])
```

# Variables de decisión

```python
for variable in modelo.variables():

    print(

        variable.name,

        "=",

        variable.varValue

    )
```

# Costo Óptimo

```python
print(

    "Costo Total =",

    value(modelo.objective)

)
```

# Interpretación de Resultados

La solución indica que:

- La Planta A envía las 40 toneladas disponibles al Centro de Distribución.
- La Planta B envía las 60 toneladas disponibles al Centro de Distribución.
- El Centro de Distribución abastece completamente a ambos clientes.

El costo mínimo obtenido para satisfacer toda la demanda es de **950 unidades monetarias**.
