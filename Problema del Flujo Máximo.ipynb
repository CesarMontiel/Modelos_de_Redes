# Introducción

Uno de los problemas clásicos de la Investigación de Operaciones es determinar la **máxima cantidad de flujo** que puede enviarse desde un nodo origen hacia un nodo destino sin exceder la capacidad de las conexiones de la red.

Este problema tiene aplicaciones en:

- Redes carreteras
- Redes ferroviarias
- Redes de distribución
- Transporte multimodal
- Redes hidráulicas
- Redes eléctricas
- Internet

A diferencia del Problema del Camino Más Corto, donde el objetivo es minimizar un costo, el Problema del Flujo Máximo busca **maximizar la cantidad transportada**.

# Conceptos básicos

## Flujo

Cantidad de unidades que circulan por un arco.

Ejemplos:

- Vehículos por hora.
- Toneladas por día.
- Contenedores.
- Paquetes.
- Personas.

---

## Capacidad

Es la cantidad máxima de flujo que puede circular por un arco.

Siempre debe cumplirse

\[
0 \leq f_{ij} \leq c_{ij}
\]

donde

- \(f_{ij}\) = flujo enviado

- \(c_{ij}\) = capacidad máxima del arco.

# Conservación del flujo

En todos los nodos intermedios debe cumplirse el principio de conservación.

Todo el flujo que entra debe salir.

\[
\sum f_{entrada}
=
\sum f_{salida}
\]

Esta condición garantiza la continuidad del flujo dentro de la red.

````python
import networkx as nx
import matplotlib.pyplot as plt
````

# Construcción de la red

La siguiente red representa un sistema de distribución de mercancías.

Cada arco posee una capacidad máxima expresada en toneladas por día.

````python
G = nx.DiGraph()

G.add_edge("A","B",capacity=15)
G.add_edge("A","C",capacity=10)

G.add_edge("B","D",capacity=8)
G.add_edge("B","E",capacity=7)

G.add_edge("C","D",capacity=5)
G.add_edge("C","E",capacity=10)

G.add_edge("D","F",capacity=10)
G.add_edge("E","F",capacity=12)
````

# Visualización de la red

````python
pos = {
    "A":(0,2),
    "B":(2,3),
    "C":(2,1),
    "D":(4,3),
    "E":(4,1),
    "F":(6,2)
}

plt.figure(figsize=(10,6))

nx.draw_networkx(
    G,
    pos,
    node_size=1200,
    node_color="lightblue",
    font_weight="bold"
)

capacidades = nx.get_edge_attributes(
    G,
    "capacity"
)

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=capacidades
)

plt.axis("off")
plt.show()
````

# Cálculo del Flujo Máximo

NetworkX implementa el algoritmo de Edmonds-Karp mediante la función:

```python
networkx.maximum_flow()
```

La función devuelve:

- El valor máximo del flujo.
- El flujo enviado por cada arco.

````python
flujo_maximo, flujo = nx.maximum_flow(
    G,
    "A",
    "F"
)
````

# Flujo máximo obtenido

````python
print("Flujo máximo =", flujo_maximo)
````

# Flujo enviado por cada arco

````python
for origen in flujo:

    print("\nNodo:", origen)

    for destino in flujo[origen]:

        print(
            f"{origen} -> {destino}:",
            flujo[origen][destino]
        )
````

# Interpretación de resultados

Cada arco indica la cantidad de flujo utilizada.

Cuando el flujo es igual a la capacidad del arco, significa que dicho arco se encuentra trabajando al máximo de su capacidad.

````python
capacidades = nx.get_edge_attributes(
    G,
    "capacity"
)

print("{:<10}{:<10}{:<10}{:<10}".format(
    "Origen",
    "Destino",
    "Flujo",
    "Capacidad"
))

print("-"*45)

for origen in flujo:

    for destino in flujo[origen]:

        print("{:<10}{:<10}{:<10}{:<10}".format(

            origen,

            destino,

            flujo[origen][destino],

            capacidades[(origen,destino)]

        ))
````

# Identificación de cuellos de botella

Los cuellos de botella corresponden a aquellos arcos donde

Flujo = Capacidad

Estos arcos limitan el flujo total de la red.

````python
print("Cuellos de botella\n")

for arco, capacidad in capacidades.items():

    origen, destino = arco

    if flujo[origen][destino] == capacidad:

        print(
            origen,
            "->",
            destino,
            "Capacidad:",
            capacidad
        )
````

# Visualización del flujo utilizado

En la siguiente figura se muestran los valores de flujo enviados por cada conexión.

````python
plt.figure(figsize=(10,6))

nx.draw_networkx(
    G,
    pos,
    node_color="lightgreen",
    node_size=1200,
    font_weight="bold"
)

labels = {}

for origen in flujo:

    for destino in flujo[origen]:

        labels[(origen,destino)] = (
            str(flujo[origen][destino]) +
            "/" +
            str(capacidades[(origen,destino)])
        )

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=labels
)

plt.axis("off")

plt.show()
````
