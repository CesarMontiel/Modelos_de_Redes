## Una red sencilla visualizada

```python
import networkx as nx
import matplotlib.pyplot as plt

# Crear una red sencilla
G = nx.Graph()
G.add_edges_from([(1, 2), (1, 3), (2, 4), (3, 4), (4, 5)])

# Dibujar la red
nx.draw(G, with_labels=True, node_color='lightblue', node_size=700, font_size=10, font_color='black')
plt.title("Red Sencilla")
plt.show()
```
